```python
print("AI Chatbot")
print("Type 'bye' to exit")

while True:
    user = input("You: ").lower()

    if user == "hello" or user == "hi":
        print("Bot: Hello! How can I help you?")
    
    elif user == "how are you":
        print("Bot: I am fine. How can I assist you?")
    
    elif user == "what is your name":
        print("Bot: I am a Python AI Chatbot.")
    
    elif user == "python":
        print("Bot: Python is a programming language used for AI, ML and data science.")
    
    elif user == "bye":
        print("Bot: Goodbye! Have a nice day.")
        break
    
    else:
        print("Bot: Sorry, I didn't understand that.")
```

```python
import nltk
import random
from nltk.chat.util import Chat, reflections

# Chat patterns and responses
pairs = [

[
r"hi|hello|hey",
["Hello! How can I help you today?",
 "Hi there! What can I do for you?",
 "Hello Rupali! Nice to meet you."]
],

[
r"what is your name ?",
["My name is Python AI Chatbot.",
 "You can call me AI Assistant."]
],

[
r"how are you ?",
["I am doing great. Thanks for asking!",
 "I am fine and ready to help you."]
],

[
r"(.*) python",
["Python is a powerful programming language used in AI, Machine Learning, and Data Science.",
 "Python is widely used for software development and automation."]
],

[
r"(.*) internship",
["Internships help students gain practical knowledge.",
 "Doing internships improves real-world experience."]
],

[
r"(.*) machine learning",
["Machine Learning is a field of Artificial Intelligence that allows computers to learn from data."]
],

[
r"bye",
["Goodbye! Have a great day.",
 "Bye! Nice talking to you."]
]

]

print("AI Chatbot is running...")
print("Type 'bye' to exit.")

chatbot = Chat(pairs, reflections)

chatbot.converse()
```

```python
import nltk
import random
import json
from datetime import datetime
from nltk.tokenize import word_tokenize

# FAQ knowledge base
knowledge_base = {
    "hello": ["Hello! Welcome to Algonive Support.", 
              "Hi there! How can I assist you today?"],
    
    "internship": ["Our internship program provides hands-on experience in Python, AI and Data Science.",
                   "Internships help students gain practical industry knowledge."],
    
    "python": ["Python is a powerful programming language widely used in AI, ML and Data Science."],
    
    "machine learning": ["Machine Learning allows computers to learn patterns from data."],
    
    "help": ["You can ask me about Python, AI, Machine Learning or Internship."]
}

# Save chat history
def save_chat(user, bot):
    data = {
        "time": str(datetime.now()),
        "user": user,
        "bot": bot
    }
    
    with open("chat_history.txt", "a") as file:
        file.write(json.dumps(data) + "\n")

# Response generator
def chatbot_response(user_input):
    
    tokens = word_tokenize(user_input.lower())
    
    for key in knowledge_base:
        if key in user_input.lower():
            response = random.choice(knowledge_base[key])
            return response
    
    return "Sorry, I don't understand. Can you rephrase your question?"

# Main chatbot loop
print("Algonive AI Support Chatbot")
print("Type 'exit' to stop the chatbot.\n")

while True:
    
    user_input = input("You: ")
    
    if user_input.lower() == "exit":
        print("Bot: Thank you for chatting. Goodbye!")
        break
    
    response = chatbot_response(user_input)
    
    print("Bot:", response)
    
    save_chat(user_input, response)
```
